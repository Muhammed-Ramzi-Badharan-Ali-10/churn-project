# 03. Modeling And Evaluation

```python
import joblib
import pandas as pd
import numpy as np
# Load the processed data
df = joblib.load("../data/processed/processed_data.pkl")
```

```python
# Split The Data Into Training And Testing Sets

from sklearn.model_selection import train_test_split

# split data
x = df.drop('Churn', axis=1) # Features
y = df['Churn'] # Target

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)

print("Train shape:", x_train.shape)
print("Test shape:", x_test.shape)
```

```python
# Build And Train Logistic Regression (Baseline Model)

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Baseline mode - Logistic Regression
log_reg = LogisticRegression(max_iter=1000)

# Train the model
log_reg.fit(x_train, y_train)

# Predict on test data
y_pred = log_reg.predict(x_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

The Logistic Regression model shows reasonable performance on the test data. The precision and recall values indicate that the model can identify churners fairly well, but there is still room for improvement. I will compare this with other models to see if I can achieve better accuracy and balanced metrics.

```python
# Build and train Decision Tree Model

from sklearn.tree import DecisionTreeClassifier

# Decision Tree Model
dt = DecisionTreeClassifier(random_state=42)

# Train the model
dt.fit(x_train, y_train)

# Predict on test data
y_pred_dt = dt.predict(x_test)

# Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred_dt))
print("\nClassification Report:\n", classification_report(y_test, y_pred_dt))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_dt))
```

The Decision Tree model performs moderately well with around 71% accuracy. It predicts non-churn customers better than churners, as seen from the lower precision and recall for class 1. The tree may be overfitting slightly, so tuning or using an ensemble model like Random Forest could improve the performance.

```python
# Build and train Random Forest Classifier

from sklearn.ensemble import RandomForestClassifier

# Random Forest Classifier Model
rf_model = RandomForestClassifier(random_state=42)

# Train the model
rf_model.fit(x_train, y_train)

# Predict on test data
rf_pred = rf_model.predict(x_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, rf_pred))
print("\nClassification Report:\n", classification_report(y_test, rf_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, rf_pred))
```

The Random Forest model generally performs better than a single Decision Tree because it reduces overfitting by combining multiple trees as we can see here. But now, I am going to do some hyperparameter tuning for achieving more **Accuracy**.

```python
# Hyperparameter Tuning for Random Forest Classifier model
from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth" : [None, 5, 10, 15],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    
}

grid_search = GridSearchCV(
    estimator = rf_model,
    param_grid = param_grid,
    cv = 3,
    scoring = "accuracy",
    n_jobs = -1
)

grid_search.fit(x_train, y_train)

print("\nBest parameters:", grid_search.best_params_)
print("Best Cross-Val Score:", grid_search.best_score_)

```

```python
# Evaluate the Best Random Forest Classifier Model

best_rf = grid_search.best_estimator_
best_rf_pred = best_rf.predict(x_test)

print("\nFinal Accuracy:", accuracy_score(y_test, best_rf_pred))
print("\nClassification Report:\n", classification_report(y_test, best_rf_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, best_rf_pred))
```

After hyperparameter tuning, the Random Forest model shows stable performance with improved generalization compared to a single Decision Tree. While the model predicts non-churn customers very well, recall for churn customers is moderate, indicating that further improvements may be achieved using advanced models or class imbalance techniques.

```python
# Build And Train Random Forest With Class Weights

rf_balanced = RandomForestClassifier(
    random_state=42,
    class_weight='balanced'
)

# Train the model
rf_balanced.fit(x_train, y_train)

# Predict on test data
y_pred_bal = rf_balanced.predict(x_test)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred_bal))
print("\nClassification Report:\n", classification_report(y_test, y_pred_bal))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_bal))
```

Accuracy is almost the same as tuned Random Forest. Churn recall did not improve (still ~0.49). So class weights alone are not enough for this dataset.
So, further improvements may be achieved using advanced models.

```python
# Install XGBoost
!pip install xgboost
```

```python
# Build And Train XGBoost Classifier

from xgboost import XGBClassifier

# XGBoost Model
xgb_model = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.08,
    scale_pos_weight = (len(y_train[y_train==0]) / len(y_train[y_train==1])), # Handle imbalance
    random_state=42,
    eval_metric='logloss'
)

# Train the model
xgb_model.fit(x_train, y_train)

# Get Churn Probabilities
y_prob_xgb = xgb_model.predict_proba(x_test)[:, 1]

# Set Custom Threshold
threshold = 0.45
y_pred_thresh = (y_prob_xgb >= threshold).astype(int)

# Evaluate
print(f"Threshold: {threshold}")
print("Accuracy:", accuracy_score(y_test, y_pred_thresh))
print("\nClassification Report:\n", classification_report(y_test, y_pred_thresh))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred_thresh))
```

By tuning the probability threshold, the model achieves a better balance between identifying churn customers and controlling false positives. A threshold of 0.45 provides strong churn recall while maintaining reasonable overall accuracy, aligning the model with real-world business objectives focused on **customer retention**.

```python
# Save The Final XGBoost Model
joblib.dump(xgb_model, "../models/final_xgboost_model.pkl")

# Save The Selected Threshold Serperately
joblib.dump(0.45, "../models/optimal_threshold.pkl")

print("Final XGBoost model and threshold saved successfully!")
```

```python
# Model Comparison Table
model_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "Random Forest (Tuned)",
        "XGBoost (Threshold = 0.45)"
    ],
    "Accuracy": [
       0.80, 
       0.71,
       0.79, 
       0.70
    ],
    "Churn Recall (Class 1)": [
        0.57,
        0.45,
        0.50,
        0.83
    ]
})

model_comparison
```

Among all models, **XGBoost** with threshold tuning provides the **highest recall** for churn customers, making it the most suitable choice for this business problem. Although its accuracy is slightly lower, the ability to identify customers at risk of churning is more valuable for **retention-focused strategies**.
