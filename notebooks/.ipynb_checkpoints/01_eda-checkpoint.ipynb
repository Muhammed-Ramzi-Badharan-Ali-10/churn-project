# 01. Exploratory Data Analysis (EDA)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv")

# Understanding the data
print(df.shape)

```

```python
print(df.head())
```

The first 5 rows give an overview of the dataset. It has customer information such as demographics (gender, SeniorCitizen, Partner, Dependents), services used (PhoneService, InternetService, StreamingTV, etc.), and financial details (Contract, PaymentMethod, MonthlyCharges, TotalCharges). The target column is **Churn**.


```python
print(df.info())
```

The dataset has 7043 entries and 21 columns.  
- Most columns are categorical.  
- **SeniorCitizen** is coded as 0 (No) and 1 (Yes) but it is wrongly stored as numeric, which will need fixing.
- **tenure** and **MonthlyCharges** are numeric features.  
- **TotalCharges** is wrongly stored as object (string) instead of numeric, which will need fixing.  
- No missing values are reported here.


```python
print(df.isnull().sum())
```

All columns currently have 0 missing values. Dataset is complete at this stage.  
But, `TotalCharges` may require conversion to numeric, which could create missing values.


```python
print(df['Churn'].value_counts(normalize=True))
```

The dataset is imbalanced: ~73.5% customers did not churn, ~26.5% churned.  
This should be considered during model training to avoid biased predictions.


```python
# Plot target variable distribution
plt.figure(figsize=(6,4))
churn_perc = df['Churn'].value_counts(normalize=True)*100
churn_perc.plot(kind="bar")
plt.title("Churn Distribution (in %)")
plt.ylabel("Percentage")
plt.show
```

From this plot also it is clearly visible that the dataset is imbalanced: ~73.5% customers did not churn, ~26.5% churned.

```python
# Plot churn by gender
plt.figure(figsize=(6,4))
sns.countplot(data= df, x= 'Churn', hue= 'gender')
plt.title("Churn by Gender")
plt.ylabel("Count")
plt.show()
```

Both genders behaved in similar fashion when it comes to migrating to another service provider/firm.
Also, both genders had almost the same proportion of customers who did not churn.



```python
# plot Churn vs SeniorCitizen
plt.figure(figsize=(6,4))
sns.countplot(data= df, x= 'SeniorCitizen', hue= 'Churn')
plt.title("Churn by SeniorCitizen Status")
plt.xlabel("SeniorCitizen (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

The plot shows that non-senior customers (0) make up the majority of the dataset, while senior customers (1) are fewer. However, the churn rate is visibly higher among seniors compared to non-seniors, indicating that older customers are more likely to leave.

```python
# Churn by Partner Status
plt.figure(figsize=(6,4))
sns.countplot(data= df, x= "Partner", hue= "Churn")
plt.title("Churn by Partner Status")
plt.show()
```

Customers **without a partner** show a higher proportion of churn compared to those **with a partner**.  
This suggests that customers with partners may have greater stability or lower likelihood of leaving the service.


```python
# Churn vs Dependents
plt.figure(figsize=(6,4))
sns.countplot(data= df, x= "Dependents", hue= "Churn")
plt.title("Churn vs Dependents")
plt.show()
```

Customers **without dependents** are more likely to churn compared to those **with dependents**.  
This indicates that customers with dependents may be more stable and less likely to leave the service.


```python
# plot Churn vs Tenure
plt.figure(figsize= (8,5))
sns.histplot(data= df, x= 'tenure', hue= 'Churn', multiple= 'stack', bins= 30)
plt.title('Churn vs Tenure')
plt.xlabel('Tenure (months)')
plt.ylabel('Count')
plt.show()
```

Churn is **higher among customers with low tenure (new customers)**, and it decreases as tenure increases.  
This suggests that customers who stay longer are less likely to churn.

```python
# Plot Churn vs PhoneSevice
sns.countplot(data= df, x= "PhoneService", hue= "Churn")
plt.title("Churn vs PhoneService")
plt.show()
```

Customers without a phone service are very few compared to those with it. Among both groups, churn rates appear similar, suggesting that having or not having a phone service does not strongly influence churn.

```python
#  Plot Churn vs MultipleLines
sns.countplot(data= df, x= "MultipleLines", hue= "Churn")
plt.title("Churn vs MultpleLines")
plt.show()
```

Customers with "No phone service" show very low churn (almost negligible).  
Among those with phone service, both "No" and "Yes" for multiple lines have similar churn rates, suggesting multiple lines does not strongly affect churn.

```python
# Plot Churn vs Internet Service
sns.countplot(data= df, x= "InternetService", hue= "Churn")
plt.title("Churn vs Internet Service")
plt.show()
```

Customers using **Fiber optic** internet have the highest churn rate compared to those with **DSL** or **No internet service**.  
This suggests that churn is strongly influenced by the type of internet service provided.

```python
# Plot Churn by Contract type
plt.figure(figsize=(6,4))
sns.countplot(data= df, x= 'Contract', hue= 'Churn')
plt.title('Churn by Contract Type')
plt.ylabel('Count')
plt.show()
```

Customers with month-to-month contracts have the highest churn rate, while those with one-year or two-year contracts are less likely to churn. This suggests that long-term contracts help retain customers.

```python
# Plot Churn by Payment Method
plt.figure(figsize=(8,5))
sns.countplot(data= df, x= 'PaymentMethod', hue= 'Churn')
plt.title('Churn by Payment Method')
plt.xlabel('Payment Method')
plt.ylabel('Count')
plt.xticks(rotation= 45)
plt.show()

```

Customers who pay through Electronic check have a much higher churn rate compared to those using bank transfer or credit card (automatic). This suggests that customers on automatic payment methods tend to stay longer, possibly due to convenience and commitment.

```python
# Plot Churn vs MonthlyCharges
plt.figure(figsize= (6,4))
sns.boxplot(data= df, x= 'Churn', y= 'MonthlyCharges')
plt.title('Churn vs MonthlyCharges')
plt.xlabel('Churn')
plt.ylabel('MonthlyCharges')
plt.show()
```

Customers who churned generally have higher monthly charges compared to those who stayed. This suggests that higher service costs might contribute to customer churn.

```python
# Plot Churn vs TotalCharges
df['TotalCharges']= pd.to_numeric(df['TotalCharges'], errors= 'coerce') # Converting 'TotalCharges' to numeric, since it was earlier stored as an object.
plt.figure(figsize= (6,4))
sns.boxplot(data= df, x= 'Churn', y= 'TotalCharges')
plt.title('Churn vs TotalCharges')
plt.show()
```

Customers who churn tend to have lower TotalCharges compared to those who stay. This suggests that customers leaving the service are likely newer or have shorter tenure, resulting in less total spending.

```python
# Plot Churn vs OnlineSecurity
plt.figure(figsize= (6,4))
sns.countplot(data= df, x= 'OnlineSecurity', hue= 'Churn')
plt.title('Churn vs OnlineSecurity')
plt.xlabel('OnlineSecurity')
plt.ylabel('Count')
plt.show()
```

Customers who do not have Online Security show a much higher churn rate compared to those who have it. This suggests that providing Online Security services may help reduce customer churn.

```python
# Plot Churn vs TechSupport
plt.figure(figsize= (6,4))
sns.countplot(data= df, x= 'TechSupport', hue= 'Churn')
plt.title('Churn vs TechSupport')
plt.show()
```

Customers without tech support are much more likely to churn compared to those who have it. Having tech support seems to increase customer retention.

```python
# Plot Correlation Heatmap
plt.figure(figsize=(25,10))
corr = df.apply(lambda x: pd.factorize(x)[0]).corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
ax = sns.heatmap(corr, mask=mask, xticklabels=corr.columns, yticklabels=corr.columns, annot=True, linewidths=.2, cmap= 'coolwarm', vmin=-1, vmax=1)
```

The heatmap shows how features are correlated with each other and with Churn.
